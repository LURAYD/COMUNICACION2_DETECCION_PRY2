```python
import numpy as np
import matplotlib.pyplot as plt


x = np.array([1+1j,1-1j,-1+1j,-1-1j])

I = np.real(x)
Q = np.imag(x)

print("Muestras I/Q:", x)

print("Muestras I:", I)

print("Muestras Q:", Q)

#MODULOS

modulos = np.abs(x)
print("Modulos:", modulos)

#potencia

P = np.mean(modulos**2)
print("Potencia:", P)


#Dibujar graficos
plt.figure(figsize=(6,6))
plt.scatter(I, Q, color='blue', label='Muestras I/Q')
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.xlim(-2, 2)
plt.ylim(-2, 2)
plt.title('Diagrama de constelación')
plt.xlabel('I (In-phase)')
plt.ylabel('Q (Quadrature)')
plt.grid()
plt.legend()
plt.show()

```
